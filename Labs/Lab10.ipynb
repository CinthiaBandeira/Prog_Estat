```python
#exercicio 1

import numpy as np
import math

n = 10**4  

# estimando por monte carlo
X = np.random.uniform(0, 1, n)

# criando uma lista e guardando os valores da função pra cada amostra
f_X = []
for x in X:
    valor = math.exp(-x**2) 
    f_X.append(valor)

# estimando a media
soma = 0
for valor in f_X:
    soma = soma + valor

I_estimado = soma / n
print(f"valor estimado da integral: {I_estimado:.4f}")

# calculando a variancia do estimador
soma_quadrados = 0
for valor in f_X:
    diferenca = valor - I_estimado
    soma_quadrados = soma_quadrados + (diferenca ** 2)

variancia_amostral = soma_quadrados / (n - 1)

variancia_estimador = variancia_amostral / n

print(f"variância do estimador:{variancia_estimador:.8f}")

#calculando IC 
z_critico = 1.96 
erro_padrao = math.sqrt(variancia_estimador)

limite_inferior = I_estimado - z_critico * erro_padrao
limite_superior = I_estimado + z_critico * erro_padrao

print(f"ic de 95%: [{limite_inferior:.4f}, {limite_superior:.4f}]")
```

```python
#exercicio 2

n = 10**4  

amostra = []

# estimando por monte carlo
while len(amostra) < n:
    u1 = np.random.uniform(0, 1)
    u2 = np.random.uniform(0, 1)

    # usando o box-muller
    r = math.sqrt(-2 * math.log(u1))
    teta = 2 * math.pi * u2

    # gerando dois valores de normais e adicionando na amostra
    z1 = r * math.cos(teta)
    z2 = r * math.sin(teta)
    
    if len(amostra) < n:
        amostra.append(z1)
    if len(amostra) < n:
        amostra.append(z2)
        
indicadora = []
for x in amostra:
    if x > 1:
        indicadora.append(1)
    else:
        indicadora.append(0)
        
# calculando o estimador
soma = 0
for valor in indicadora:
    soma = soma + valor

p_estimado = soma / n
print(f"valor estimado: {p_estimado:.4f}")

# calculando a variancia do estimador
variancia_amostral = p_estimado * (1 - p_estimado)

variancia_estimador = variancia_amostral / n

print(f"variância do estimador: {variancia_estimador:.8f}")

#calculando ic de 95%
z_critico = 1.96 
erro_padrao = math.sqrt(variancia_estimador)

limite_inferior = p_estimado - z_critico * erro_padrao
limite_superior = p_estimado + z_critico * erro_padrao

print(f"ic de 95%: [{limite_inferior:.4f}, {limite_superior:.4f}]")
```

```python
#exercicio 3

n = 10**4 

# estimando por monte carlo
X = np.random.uniform(0, 1, n)
Y = np.random.uniform(0, 1, n)

indicadora = []

for i in range(n):
    if (X[i]**2 + Y[i]**2) <= 1:
        indicadora.append(1)
    else:
        indicadora.append(0)

# estimando o valor de pi
soma = 0
for valor in indicadora:
    soma = soma + valor

p_estimado = soma / n

pi_estimado = 4 * p_estimado

print(f"valor estimado de pi: {pi_estimado:.4f}")

# calculando variancia do estimador
variancia_amostral = p_estimado * (1 - p_estimado)

variancia_p = variancia_amostral / n

variancia_estimador_pi = 16 * variancia_p

print(f"variância do estimador: {variancia_estimador_pi:.8f}")

# calculando ic 95%
z_critico = 1.96 
erro_padrao_pi = math.sqrt(variancia_estimador_pi)

limite_inferior = pi_estimado - z_critico * erro_padrao_pi
limite_superior = pi_estimado + z_critico * erro_padrao_pi

print(f"ic de 95%: [{limite_inferior:.4f}, {limite_superior:.4f}]")
```

```python
#exercicio 4

n = 10**4  

# estimando a média por monte carlo
X = np.random.uniform(0, 1, n)

soma = 0
for valor in X:
    soma = soma + valor

media_estimada = soma / n
print(f"E[X]: {media_estimada:.4f}")

# calculando a variancia do estimador
soma_quadrados = 0
for valor in X:
    diferenca = valor - media_estimada
    soma_quadrados = soma_quadrados + (diferenca ** 2)

variancia_estimada = soma_quadrados / (n - 1)
print(f"Var(X): {variancia_estimada:.4f}")

variancia_estimador = variancia_estimada / n

print(f"Variância do estimador da média: {variancia_estimador:.8f}")

#ic de 95%
z_critico = 1.96 
erro_padrao = math.sqrt(variancia_estimador)

limite_inferior = media_estimada - z_critico * erro_padrao
limite_superior = media_estimada + z_critico * erro_padrao

print(f"ic de 95% para E[X]: [{limite_inferior:.4f}, {limite_superior:.4f}]")
```

```python
#exercicio 7

n = 10**4  

amostra = []
# gerando as amostras da exponencial por inversão
for i in range(n):
    u = np.random.uniform(0, 1)
    x = -math.log(u)
    amostra.append(x)

# aplpicando a função em cada amostra
f_X = []
for x in amostra:
    f_X.append(1 / (1 + x))

# calculando o estimador por monte carlo
soma = 0
for valor in f_X:
    soma = soma + valor

I_estimado = soma / n
print(f"valor estimado: {I_estimado:.4f}")

# calculando a variancia do estimador
soma_quadrados = 0
for valor in f_X:
    diferenca = valor - I_estimado
    soma_quadrados = soma_quadrados + (diferenca ** 2)

variancia_amostral = soma_quadrados / (n - 1)
variancia_estimador = variancia_amostral / n

print(f"variância do estimador: {variancia_estimador:.8f}")

#ic 95%
z_critico = 1.96 
erro_padrao = math.sqrt(variancia_estimador)

limite_inferior = I_estimado - z_critico * erro_padrao
limite_superior = I_estimado + z_critico * erro_padrao

print(f"ic de 95%: [{limite_inferior:.4f}, {limite_superior:.4f}]")
```

```python
#exercicio 8

n = 10**4  
L = 20     

# estimando por monte carlo
X = np.random.uniform(0, L, n)

# estimativa da integral
f_X = []
for x in X:
    if x == 0:
        f_X.append(1.0) 
    else:
        f_X.append(math.sin(x) / x)

# estimando a media
soma = 0
for valor in f_X:
    soma = soma + valor

media_estimada = soma / n

I_estimado = L * media_estimada
print(f"valor estimado: {I_estimado:.4f}")

# calculando variancia do estimador
soma_quadrados = 0
for valor in f_X:
    diferenca = valor - media_estimada
    soma_quadrados = soma_quadrados + (diferenca ** 2)

variancia_amostral = soma_quadrados / (n - 1)
variancia_media = variancia_amostral / n

variancia_estimador = (L ** 2) * variancia_media

print(f"variância do estimador: {variancia_estimador:.6f}")

#ic de 95%
z_critico = 1.96 
erro_padrao = math.sqrt(variancia_estimador)

limite_inferior = I_estimado - z_critico * erro_padrao
limite_superior = I_estimado + z_critico * erro_padrao

print(f"ic de 95%: [{limite_inferior:.4f}, {limite_superior:.4f}]")
```
