```python
# distribuicoes = {
#     "Normal": {"params": {"loc": 0, "scale": 1}, "dados": None, "estatisticas": None},
#     "Exponencial": {"params": {"scale": 2}, "dados": None, "estatisticas": None},
#     "Uniforme": {"params": {"low": -1, "high": 1}, "dados": None, "estatisticas": None}
# }

#Passo 1: criando dicionario principal das distribuições
dados = None

estatisticas = None

parametros_normal = {"params": {"loc": 0, "scale": 1}, "dados": dados, "estatisticas": estatisticas}

parametros_exponencial = {"params": {"scale": 2}, "dados": dados, "estatisticas": estatisticas}

parametros_uniforme = {"params": {"low": -1, "high": 1}, "dados": dados, "estatisticas": estatisticas}


distribuicoes = {"Normal": parametros_normal, "Exponencial": parametros_exponencial, "Uniforme": parametros_uniforme}

print(distribuicoes)
```

```python
#Passo 2: importando as bibliotecas

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Passo 3: gerando amostras aleatorias das distribuições

amostra_normal = np.random.normal(loc=0, scale=1, size=(1000))

amostra_exponencial = np.random.exponential(scale=2, size=(1000))

amostra_uniforme = np.random.uniform(low=-1, high=1, size=(1000))

distribuicoes["Normal"]["dados"] = amostra_normal
distribuicoes["Exponencial"]["dados"] = amostra_exponencial
distribuicoes["Uniforme"]["dados"] = amostra_uniforme


```

```python
#Passo 4: criando as estatisticas básicas (média, mediana, variância e dp) das distribuições

#estatisticas normal
media_normal = np.mean(amostra_normal)
mediana_normal = np.median(amostra_normal)
variancia_normal = np.var(amostra_normal)
dp_normal = np.std(amostra_normal)

estatisticas_normal = {"Media":media_normal, "Mediana":mediana_normal, "Variancia":variancia_normal, "DP":dp_normal}
distribuicoes["Normal"]["estatisticas"] = estatisticas_normal

#estatisticas exponencial
media_exp = np.mean(amostra_exponencial)
mediana_exp = np.median(amostra_exponencial)
variancia_exp = np.var(amostra_exponencial)
dp_exp = np.std(amostra_exponencial)

estatisticas_exp = {"Media":media_exp, "Mediana":mediana_exp, "Variancia":variancia_exp, "DP":dp_exp}
distribuicoes["Exponencial"]["estatisticas"] = estatisticas_exp

#estatisticas uniforme
media_uniforme = np.mean(amostra_uniforme)
mediana_uniforme = np.median(amostra_uniforme)
variancia_uniforme = np.var(amostra_uniforme)
dp_uniforme = np.std(amostra_uniforme)

estatisticas_uniforme = {"Media":media_uniforme, "Mediana":mediana_uniforme, "Variancia":variancia_uniforme, "DP":dp_uniforme}
distribuicoes["Uniforme"]["estatisticas"] = estatisticas_uniforme
```

```python
#Passo 5: plotando os histogramas

#hit normal
histograma_normal = plt.hist(distribuicoes["Normal"]["dados"], color= "violet")
plt.title("Distribuição Normal")
plt.show()

#hist exp
histograma_exp = plt.hist(distribuicoes["Exponencial"]["dados"], color = "violet")
plt.title("Distribuição Exponencial")
plt.show()

#hist uniforme
histograma_uniforme = plt.hist(distribuicoes["Uniforme"]["dados"], color = "violet")
plt.title("Distribuição Uniforme")
plt.show()
```

```python
#Passo 6: criando a tabela comparativa

tabela_distribuicoes = pd.DataFrame({"Normal": distribuicoes["Normal"]["estatisticas"], "Exponencial": distribuicoes["Exponencial"]["estatisticas"], "Uniforme": distribuicoes["Uniforme"]["estatisticas"]})
tabela_distribuicoes
```

```python
#Passo 7: repetindo os passos anteriores para uma nova distribuição

parametros_binomial = {"params": {"n":10, "p":0.5}, "dados": dados, "estatisticas": estatisticas}

distribuicoes["Binomial"] = parametros_binomial

amostra_binomial = np.random.binomial(n=10, p=0.5, size=(1000))

distribuicoes["Binomial"]["dados"] = amostra_binomial

#estatisticas binomial
media_binomial = np.mean(amostra_binomial)
mediana_binomial = np.median(amostra_binomial)
variancia_binomial = np.var(amostra_binomial)
dp_binomial = np.std(amostra_binomial)

estatisticas_binomial = {"Media": media_binomial, "Mediana": mediana_binomial, "Variancia": variancia_binomial, "DP": dp_binomial}
distribuicoes["Binomial"]["estatisticas"] = estatisticas_binomial

tabela_distribuicoes["Binomial"] = distribuicoes["Binomial"]["estatisticas"]
tabela_distribuicoes
```

```python
#Passo 8: atualizando a tabela com valores de minimo, maximo e quartis

#Normal
min_normal = np.min(distribuicoes["Normal"]["dados"])
max_normal = np.max(distribuicoes["Normal"]["dados"])
quartis_normal = np.percentile(distribuicoes["Normal"]["dados"], q = [25,50,75])

estatisticas_normal["Minimo"] = min_normal
estatisticas_normal["Maximo"] = max_normal
estatisticas_normal["Quartis"] = quartis_normal
distribuicoes["Normal"]["estatisticas"] = estatisticas_normal

#Exponencial
min_exp = np.min(distribuicoes["Exponencial"]["dados"])
max_exp = np.max(distribuicoes["Exponencial"]["dados"])
quartis_exp = np.percentile(distribuicoes["Exponencial"]["dados"], q = [25,50,75])

estatisticas_exp["Minimo"] = min_exp
estatisticas_exp["Maximo"] = max_exp
estatisticas_exp["Quartis"] = quartis_exp
distribuicoes["Exponencial"]["estatisticas"] = estatisticas_exp

#Uniforme
min_uniforme = np.min(distribuicoes["Uniforme"]["dados"])
max_uniforme = np.max(distribuicoes["Uniforme"]["dados"])
quartis_uniforme = np.percentile(distribuicoes["Uniforme"]["dados"], q = [25,50,75])

estatisticas_uniforme["Minimo"] = min_uniforme
estatisticas_uniforme["Maximo"] = max_uniforme
estatisticas_uniforme["Quartis"] = quartis_uniforme
distribuicoes["Uniforme"]["estatisticas"] = estatisticas_uniforme

#Binomial
min_binomial = np.min(distribuicoes["Binomial"]["dados"])
max_binomial = np.max(distribuicoes["Binomial"]["dados"])
quartis_binomial = np.percentile(distribuicoes["Binomial"]["dados"], q = [25,50,75])

estatisticas_binomial["Minimo"] = min_binomial
estatisticas_binomial["Maximo"] = max_binomial
estatisticas_binomial["Quartis"] = quartis_binomial
distribuicoes["Binomial"]["estatisticas"] = estatisticas_binomial


tabela_distribuicoes = pd.DataFrame({"Normal": distribuicoes["Normal"]["estatisticas"], 
                                     "Exponencial": distribuicoes["Exponencial"]["estatisticas"], 
                                     "Uniforme": distribuicoes["Uniforme"]["estatisticas"],
                                    "Binomial":distribuicoes["Binomial"]["estatisticas"]})
tabela_distribuicoes



```

```python
#Passo 9: plotando o boxplot

dados_gerais = [distribuicoes["Normal"]["dados"], 
                distribuicoes["Exponencial"]["dados"], 
                distribuicoes["Uniforme"]["dados"], 
                distribuicoes["Binomial"]["dados"]]

nomes = ["Normal", "Exponencial", "Uniforme", "Binomial"]

plt.boxplot(dados_gerais, tick_labels=nomes, patch_artist=True)
plt.title("Boxplot das distribuições")
plt.ylabel("Dados")
plt.show()
```

```python
#Passo 10: calculando médias teoricas e comparando com as amostrais

import numpy as np

normal = np.random.normal(loc=0.0, scale=1.0, size=1000)
media_teorica_normal = np.mean(normal)
print(media_teorica_normal)

lambd = 1.5
escala = 1 / lambd
exponencial = np.random.exponential(scale=escala, size=5)
media_teorica_exp = np.mean(exponencial)
print(media_teorica_exp)

uniforme = np.random.uniform(low=0.0, high=5.0, size=5)
media_teorica_uniforme = np.mean(uniforme)
print(media_teorica_uniforme)

binomial = np.random.binomial(n=10, p=0.5, size=1000)
media_teorica_bin = np.mean(binomial)
print(media_teorica_bin)

medias_amostrais = [distribuicoes["Normal"]["estatisticas"]["Media"], 
                distribuicoes["Exponencial"]["estatisticas"]["Media"], 
                distribuicoes["Uniforme"]["estatisticas"]["Media"], 
                distribuicoes["Binomial"]["estatisticas"]["Media"]]

medias_teoricas = [media_teorica_normal,
                 media_teorica_exp, 
                 media_teorica_uniforme,
                 media_teorica_bin]

nome_medias = ["Normal", "Exponencial", "Uniforme", "Binomial"]

x = np.arange(len(nome_medias))
largura = 0.35

figura, grafico = plt.subplots(figsize=(8, 5))

grafico.bar(x - largura/2, medias_teoricas, largura, label = 'Teórica', color = 'violet')
grafico.bar(x + largura/2, medias_amostrais, largura, label = 'Amostra', color = 'blue')

grafico.set_ylabel('Valores das Médias')
grafico.set_title('Compração entre: Médias Teóricas e Amostrais')
grafico.set_xticks(x)
grafico.set_xticklabels(nome_medias)
grafico.legend()

plt.show()
```
