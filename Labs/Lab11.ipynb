```python
#exercicio 1

import numpy as np
import math

n = 10**4     
k = 3         
lambida = 2       
valor_teorico = 1.174982

#gerando a gamma via soma de exponenciais
amostra = []

while len(amostra) < n:
    soma_exponenciais = 0
    
    for j in range(k):
        u = np.random.uniform(0, 1)
        exponencial = -math.log(u) / lambida
        
        soma_exponenciais = soma_exponenciais + exponencial
        
    amostra.append(soma_exponenciais)

# calculando mu via monte carlo
f_X = []
for x in amostra:
    f_X.append(math.sqrt(x))

soma = 0
for valor in f_X:
    soma = soma + valor

mu_estimado = soma / n
print(f"o valor teórico é: {valor_teorico} e o valor estimado é: {mu_estimado:.6f}")

#calculando a variancia do estimador
soma_quadrados = 0
for valor in f_X:
    diferenca = valor - mu_estimado
    soma_quadrados = soma_quadrados + (diferenca ** 2)

variancia_amostral = soma_quadrados / (n - 1)

variancia_estimador = variancia_amostral / n

print(f"variância do estimador: {variancia_estimador:.6f}")

# calculando ic de 95%
z_critico = 1.96
erro_padrao = math.sqrt(variancia_estimador)

limite_inferior = mu_estimado - z_critico * erro_padrao
limite_superior = mu_estimado + z_critico * erro_padrao

print(f"ic de 95%: [{limite_inferior:.6f}, {limite_superior:.6f}]")
```

```python
#exercicio 2

n = 10**4    
alpha = 2      
beta = 5       
x = 0.4   
valor_teorico = 0.23328

#definindo a função beta
def funcao_beta(x):
    if 0 <= x <= 1:
        return 30 * x * (1 - x)**4
    return 0

moda = (alpha - 1) / (alpha + beta - 2)
c = funcao_beta(moda)

#gerando beta via aceitação-rejeição
amostra = []

while len(amostra) < n:
    y = np.random.uniform(0, 1)
    u = np.random.uniform(0, 1)
    if u <= funcao_beta(y) / c:
        amostra.append(y)

#estimando p via monte carlo
indicadora = []
for t in amostra:
    if t > x:
        indicadora.append(1)
    else:
        indicadora.append(0)
        
soma = 0
for valor in indicadora:
    soma = soma + valor

p_estimado = soma / n

print(f"o valor teórico p é: {valor_teorico} e o valor estimado é: {p_estimado:.6f}")

#calculando a variancia do estimador
variancia_amostral = p_estimado * (1 - p_estimado)

variancia_estimador = variancia_amostral / n

print(f"a variância do estimador é: {variancia_estimador:.8f}")

#ic de 95%
z_critico = 1.96 
erro_padrao = math.sqrt(variancia_estimador)

limite_inferior = p_estimado - z_critico * erro_padrao
limite_superior = p_estimado + z_critico * erro_padrao

print(f"ic de 95%: [{limite_inferior:.6f}, {limite_superior:.6f}]")
```

```python
#exercicio 3

n = 10**4      
mu = 1         
dp = 2      
valor_teorico = 8.0

#amostra normal via box-muller
amostra = []

while len(amostra) < n:
    u1 = np.random.uniform(0, 1)
    u2 = np.random.uniform(0, 1)
    
    r = math.sqrt(-2 * math.log(u1))
    teta = 2 * math.pi * u2
    
    z1 = r * math.cos(teta)
    z2 = r * math.sin(teta)
    
    x1 = mu + dp * z1
    x2 = mu + dp * z2
    
    if len(amostra) < n:
        amostra.append(x1)
    if len(amostra) < n:
        amostra.append(x2)

#estimando I via monte carlo
f_X = []
for x in amostra:
    valor = (x**2) + (2 * x) + 1
    f_X.append(valor)

soma = 0
for valor in f_X:
    soma = soma + valor

I_estimado = soma / n
print(f"o valor teórico I é: {valor_teorico} e o valor estimado é: {I_estimado:.4f}")

#calculando a variancia do estimador
soma_quadrados = 0
for valor in f_X:
    diferenca = valor - I_estimado
    soma_quadrados = soma_quadrados + (diferenca ** 2)

variancia_amostral = soma_quadrados / (n - 1)

variancia_estimador = variancia_amostral / n

print(f"a vriância do estimador é: {variancia_estimador:.6f}")

#ic de 95%
z_critico = 1.96 
erro_padrao = math.sqrt(variancia_estimador)

limite_inferior = I_estimado - z_critico * erro_padrao
limite_superior = I_estimado + z_critico * erro_padrao

print(f"ic de 95%: [{limite_inferior:.4f}, {limite_superior:.4f}]")
```

```python
#exercicio 4

n = 10**4      
valor_teorico = 0.75

#gerando a gamma (amostra X) e a beta (amostra Y)
amostra_X = []
amostra_Y = []

for i in range(n):
    soma_exponenciais = 0
    for j in range(2): 
        u = np.random.uniform(0, 1)
        soma_exponenciais += -math.log(u)
    amostra_X.append(soma_exponenciais)

    u_y = np.random.uniform(0, 1)
    y = -math.log(u_y)
    amostra_Y.append(y)

#estimando p via  monte carlo
indicadora = []
for i in range(n):
    if amostra_X[i] > amostra_Y[i]:
        indicadora.append(1)
    else:
        indicadora.append(0)

soma = 0
for valor in indicadora:
    soma = soma + valor

p_estimado = soma / n
print(f"o valor teórico p é: {valor_teorico} e o valor estimado é:{p_estimado: .4f}")

#calculando a variancia do estimador
variancia_amostral = p_estimado * (1 - p_estimado)

variancia_estimador = variancia_amostral / n

print(f"a variância do estimador é: {variancia_estimador:.6f}")

#ic de 95%
z_critico = 1.96 
erro_padrao = math.sqrt(variancia_estimador)

limite_inferior = p_estimado - z_critico * erro_padrao
limite_superior = p_estimado + z_critico * erro_padrao

print(f"ic de 95%: [{limite_inferior:.4f}, {limite_superior:.4f}]")    
```

```python
#exercicio 5

n = 10**4     
alpha = 3
beta = 4
valor_teorico = 0.02598262

#função beta
def funcao_beta(x):
    if 0 <= x <= 1:
        return 60 * (x**2) * ((1 - x)**3)
    return 0

moda = (alpha - 1) / (alpha + beta - 2)
c = funcao_beta(moda)

#gerando beta via aceitação reijeição
amostra = []

while len(amostra) < n:
    y = np.random.uniform(0, 1)
    u = np.random.uniform(0, 1)
    
    if u <= funcao_beta(y) / c:
        amostra.append(y)

#estimando I via monte carlo
f_X = []
for x in amostra:
    f_X.append(math.exp(x))

soma = 0
for valor in f_X:
    soma = soma + valor

media_exponencial = soma / n

constante_B = 1 / 60
I_estimado =constante_B * media_exponencial

print(f"o valor teórico I é: {valor_teorico} e o valor estimado é: {I_estimado:.8f}")

#calculando a variancia do estimador
soma_quadrados = 0
for valor in f_X:
    diferenca = valor - media_exponencial
    soma_quadrados = soma_quadrados + (diferenca ** 2)

variancia_amostral = soma_quadrados / (n - 1)

variancia_media = variancia_amostral / n

variancia_estimador = (constante_B ** 2) * variancia_media

print(f"a variância do estimador é: {variancia_estimador:.10f}")

#ic de 95%
z_critico = 1.96 
erro_padrao = math.sqrt(variancia_estimador)

limite_inferior = I_estimado - z_critico * erro_padrao
limite_superior = I_estimado + z_critico * erro_padrao

print(f"ic de 95%: [{limite_inferior:.8f}, {limite_superior:.8f}]")
```

```python
#exercicio 6

n = 10**4   
x = 5     
valor_teorico = 0.082085

#gerando amostras via box-muller
amostra_X = []
amostra_Y = []

while len(amostra_X) < n:
    u1 = np.random.uniform(0, 1)
    u2 = np.random.uniform(0, 1)
    
    r = math.sqrt(-2 * math.log(u1))
    teta = 2 * math.pi * u2
    
    z1 = r * math.cos(teta)
    z2 = r * math.sin(teta)
    
    if len(amostra_X) < n:
        amostra_X.append(z1)
        amostra_Y.append(z2)

#estimando p via monte carlo
indicadora = []
for i in range(n):
    distancia_quadrado = (amostra_X[i]**2) + (amostra_Y[i]**2)
    if distancia_quadrado > x:
        indicadora.append(1)
    else:
        indicadora.append(0)

soma = 0
for valor in indicadora:
    soma = soma + valor

p_estimado = soma / n
print(f"o valor teórico p é: {valor_teorico} e o valor estimado é: {p_estimado:.6f}")

#calculando a variancia do estimador
variancia_amostral = p_estimado * (1 - p_estimado)

variancia_estimador = variancia_amostral / n

print(f"a variância do estimador é: {variancia_estimador:.8f}")

#ic de 95%
z_critico = 1.96 
erro_padrao = math.sqrt(variancia_estimador)

limite_inferior = p_estimado - z_critico * erro_padrao
limite_superior = p_estimado + z_critico * erro_padrao

print(f"ic de 95%: [{limite_inferior:.6f}, {limite_superior:.6f}]")
```

```python
#exercicio 7

n = 10**4     
variaveis_X = 5   
lambida = 2       
valor_teorico = 1.141667

amostra = []

while len(amostra) < n:
    variaveis = []
    for i in range(variaveis_X):
        u = np.random.uniform(0, 1)
        x = -math.log(u) / lambida
        variaveis.append(x)
        
    maximo_atual = max(variaveis)
    amostra.append(maximo_atual)

#estimando E[M] via monte carlo
soma = 0
for valor in amostra:
    soma = soma + valor

esperanca_estimada = soma / n
print(f"o valor teórico E[M] é: {valor_teorico} e o valor estimado é: {esperanca_estimada:.6f}")

#calculando a variancia do estimador
soma_quadrados = 0
for valor in amostra:
    diferenca = valor - esperanca_estimada
    soma_quadrados = soma_quadrados + (diferenca ** 2)

variancia_amostral = soma_quadrados / (n - 1)

variancia_estimador = variancia_amostral / n

print(f"a variância do estimador é: {variancia_estimador:.8f}")

#ic de 95%
z_critico = 1.96 
erro_padrao = math.sqrt(variancia_estimador)

limite_inferior = esperanca_estimada - z_critico * erro_padrao
limite_superior = esperanca_estimada + z_critico * erro_padrao

print(f"ic de 95%: [{limite_inferior:.6f}, {limite_superior:.6f}]")
```

```python
#exercicio 8

n = 10**4    
valor_teorico_mu = 1.648721
valor_teorico_p = 0.0106511

#estimando mu via monte carlo
amostra_X = []

while len(amostra_X) < n:
    # Geramos duas Uniformes(0,1)
    u1 = np.random.uniform(0, 1)
    u2 = np.random.uniform(0, 1)
    
    # box-muller para z1 e z2 ~ N(0,1)
    r = math.sqrt(-2 * math.log(u1))
    theta = 2 * math.pi * u2
    
    z1 = r * math.cos(theta)
    z2 = r * math.sin(theta)
    
    if len(amostra_X) < n:
        amostra_X.append(math.exp(z1))
    if len(amostra_X) < n:
        amostra_X.append(math.exp(z2))

soma = 0
for valor in amostra_X:
    soma = soma + valor

mu_estimado = soma / n

print(f"o valor teórico p é: {valor_teorico_mu} e o valor estimado é: {mu_estimado:.6f}")

#calculando a variancia do estimador mu
soma_quadrados_mu = 0
for valor in amostra_X:
    diferenca = valor - mu_estimado
    soma_quadrados_mu = soma_quadrados_mu + (diferenca ** 2)

variancia_amostral_mu = soma_quadrados_mu / (n - 1)
variancia_estimador_mu = variancia_amostral_mu / n

print(f"a variância do estimador mu é: {variancia_estimador_mu:.8f}")

#ic de 95% para mu
z_critico = 1.96
erro_padrao_mu = math.sqrt(variancia_estimador_mu)
limite_inferior_mu = mu_estimado - z_critico * erro_padrao_mu
limite_superior_mu = mu_estimado + z_critico * erro_padrao_mu

print(f"ic de 95%: [{limite_inferior_mu:.6f}, {limite_superior_mu:.6f}]")

#estimando p via monte carlo
indicadora = []
for x in amostra_X:
    if x > 10:
        indicadora.append(1)
    else:
        indicadora.append(0)

soma_p = 0
for valor in indicadora:
    soma_p = soma_p + valor

p_estimado = soma_p / n

print(f"o valor teórico p é: {valor_teorico_p} e o valor estimado é: {p_estimado:.6f}")

#calculando a variância do estimador para p (Bernoulli)
variancia_amostral_p = p_estimado * (1 - p_estimado)
variancia_estimador_p = variancia_amostral_p / n

print(f"a variância do estimador p é: {variancia_estimador_p:.8f}")

#ic de 95% para p
erro_padrao_p = math.sqrt(variancia_estimador_p)
limite_inferior_p = p_estimado - z_critico * erro_padrao_p
limite_superior_p = p_estimado + z_critico * erro_padrao_p

print(f"ic de 95%: [{limite_inferior_p:.6f}, {limite_superior_p:.6f}]")
```
